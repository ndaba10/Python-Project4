<h3>Importing libraries<h3>

```python
import pandas as pd 
import seaborn as sns 

import matplotlib 
import matplotlib.pyplot as plt 
plt.style.use('ggplot')
from matplotlib.pyplot import figure
import numpy as np

%matplotlib inline
matplotlib.rcParams['figure.figsize'] = (12,8) # adjust the configuration of the plots we will create
```

<h3>Read in the data<h3>

```python
df = pd.read_csv(r'movies.csv')
```

<h3>Look at data<h3>

```python
df.head()
```

<h3>checking for missing data<h3>

```python
for col in df.columns:
    pct_missing = np.mean(df[col].isnull())
    print ('{}- {}%'.format(col, pct_missing))
```

<h3>Data type for columns<h3>

```python
df.dtypes
```

<h3>Changing data types for columns<h3>

```python
df['budget'] = df['budget'].astype('int64')

df['gross'] = df['gross'].astype('int64')
```

<h3>Create a year correct column<h3>

```python
df['yearcorrect'] = df['year'].astype(str).str[:4]

df 
```

<H3>sORTING BY GROSS<H3>

```python
df.sort_values(by=['gross'], inplace=False , ascending=False)
```

<h3>Dropping duplicates<h3>

<h3> correlation<h3>

<h3>Scatter plot with budfget vs gross<h3>

```python
plt.scatter(x=df['budget'], y=df['gross'])
plt.title('Budget vs Gross Earnings')
plt.xlabel('Gross Earnings')
plt.ylabel('Budget for film')
```

<h3> Plot budget vs gross using seaborn<h3>

```python
sns.regplot(x='budget', y='gross', data=df , scatter_kws={"color":"red"} , line_kws={"color":"blue"})
```

<h3>Looking at correlation<h3>

```python
df.corr()  #peason , kendall ,spearman
```

```python
correlation_matrix = df.corr(method='pearson')
sns.heatmap(correlation_matrix, annot=True)
plt.title('Correlation Matric for numeric features')
plt.xlabel('Movie Features')
plt.ylabel('Movie Features')
plt.show()
```

<h3>Looking at company<h3>

```python
df_numerised = df 

for col_name in df_numerised.columns:
    if(df_numerised[col_name].dtype == 'object'):
        df_numerised[col_name] = df_numerised[col_name].astype('category')
        df_numerised[col_name] = df_numerised[col_name].cat.codes

df_numerised
```

<h3>Corralatrion heatmap<h3>

```python
correlation_matrix = df_numerised .corr(method='pearson')
sns.heatmap(correlation_matrix, annot=True)
plt.title('Correlation Matric for numeric features')
plt.xlabel('Movie Features')
plt.ylabel('Movie Features')
plt.show()
```

<h3>Unstacking the correlation<h3>

```python
corr_mat = df_numerised.corr()
corr_pairs = corr_mat.unstack()
corr_pairs
```
